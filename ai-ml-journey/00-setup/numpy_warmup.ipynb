Practice array creation, indexing, vectorization, and basic mathematical operations.

```python
import numpy as np
print(f"NumPy version: {np.__version__}")
```

## 1. Array Creation & Attributes

```python
# Create a 1D array and a 2D matrix
arr_1d = np.array([1, 2, 3, 4, 5])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Shape:", arr_1d.shape)
print("2D Shape:", arr_2d.shape)
```

## 2. Basic Operations & Vectorization

```python
# Vectorized element-wise operations
squared = arr_1d ** 2
scaled = arr_1d * 10
dot_product = np.dot(arr_1d, arr_1d)

print("Squared:", squared)
print("Scaled:", scaled)
print("Dot product with itself:", dot_product)
```
